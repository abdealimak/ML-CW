```python
import numpy as np
marks = np.array([65,78,90,82])

print("Array:",marks)
print("Mean:",np.mean(marks))
print("Median:",np.median(marks))
print("Standard Deviation:",np.std(marks))
print("Maximum:",np.max(marks))
print("Minimum:",np.min(marks))

# Add 5 bonus marks to every student
update_marks = marks +5
print("Updated Marks:",update_marks)
```

```python
import numpy as np
marks = np.array([
    [75, 80, 65],
    [60, 70, 65],
    [90, 95, 92],
    [55, 60, 58]
])

print("Student marks:", marks)
print("Shape:", marks.shape)
print("Total values:", marks.size)
```

```python
import numpy as np

marks = np.array([65,78,90,55,82])

print("First student:",marks[0])
print("Third student:",marks[2])
print("Last student:",marks[-1])

print("First three students:",marks[0:3])
print("First second students:",marks[1:])
print("Every second students:",marks[::3])
```

```python
import pandas as pd

data = {
    "Student": ["Amit", "Priya", "Rahul", "Sneha", "Neha"],
    "Marks": [65, 78, 90, 55, 82],
    "Grade": ["B", "A", "A", "C", "B"]
}


df = pd.DataFrame(data)
print(df)
print("\nFirst fiive rows:")
print(df.head())
print("\nDataset information:")
print(df.info())
```

```python
import pandas as pd

data = {
    'Student': ["Amit", "Priya", "Rahul", "Sneha", "Neha"],
    "StudyHours": [2,5, None, 7,6],
    "Attendance": [60, 85, 75,None,90],
    "Marks": [45, 78, 90,55, None]
}

df = pd.DataFrame(data)
print("Missing Values in each column")
print(df.isnull().sum())

print("\nTotal missing values:")
print(df.isnull().sum().sum())
```

```python
import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr", "May"]
sales = [100,120,150,170,190]

plt.plot(months,sales)
plt.xlabel("Months")
plt.ylabel("Sales")
plt.title("Monthly Sales")
plt.show()

```

```python
students = ["A", "B", "C", "D"]
marks = [85,75,92,80]

plt.bar(students,marks)
plt.xlabel("Students")
plt.ylabel("Marks")
plt.title("Student Marks")
plt.show()
```

```python
courses = ["Python", "Java", "SQL", "C++"]
students = [150,120,90,80]

plt.barh(courses, students)
plt.xlabel("Students")
plt.ylabel("Courses")
plt.title("Course Enrollment")
plt.show()
```

```python
hours = [1,2,3,4,5,6]
marks = [35,45,55,65,75,88]

plt.scatter(hours, marks)
plt.xlabel("Study Hours")
plt.ylabel("Marks")
plt.title("Study Hours vs Marks")
plt.show()
```

```python
marks = [35,40,45,50,55,60,62,65,70,
         72,75,80,82,85,90,92,95]

plt.hist(marks, bins=20)
plt.xlabel("Marks")
plt.ylabel("Frequency")
plt.title("Marks Distribution")
plt.show()
```

```python
activities = ["Study", "Sleep", "Games", "Others"]
hours = [8,8,4,4]

plt.pie(hours, labels=activities, autopct="%1.1f%%")
plt.title("Activity Hours")
plt.show()
```

```python
marks = [45,55,60,62,65,68,70,
         72,75,80,82,85,90,92,95,150]

plt.boxplot(marks)
plt.xlabel("Marks")
plt.ylabel("Frequency")
plt.title("Marks Distribution")
plt.show()
```
